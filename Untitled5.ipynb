```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\LENOVO\Downloads\Customer_Purchase.csv")
```

```python
le = LabelEncoder()
df['Previous_Purchase'] = le.fit_transform(df['Previous_Purchase'])
df['Buy'] = le.fit_transform(df['Buy'])

X = df[['Age', 'Salary', 'Previous_Purchase']]
y = df['Buy']

model = DecisionTreeClassifier()
model.fit(X, y)
```

```python
plt.figure(figsize=(14,8))
plot_tree(
    model,
    feature_names=['Age','Salary','Previous_Purchase'],
    class_names=['Not Buy','Buy'],
    filled=True
)

plt.show()
```
